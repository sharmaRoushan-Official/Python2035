### 17. Generator in Python

Q. What is a Generator?

A generator is a special function that returns an iterator and provides values one at a time using yield instead of return.


📌 Define
> A generator generates values one by one,only when needed.

#### Why Generator are needed?
❌ Generator [list]

```python
### user-defined function
def get_numbers():
    print("hello world")
get_numbers()
```

```python
### user-defined function
def get_numbers():
    # print("hello world")
    yield "Hello world"
get_numbers()
```

```python
### user-defined function
def get_numbers():
    # print("hello world")
    yield "Hello world"
var = get_numbers()
print(next(var))
```

❌ Without Generator [List]


```python
def get_numbers():
    return [1,2,3,4,5]
get_numbers()
```

✅️ With Generator[List]

```python
def get_numbers():
    # yield [1,2,3,4,5]
    for i in range(1,6):
        yield i
get_numbers()
```

```python
x = get_numbers()
print(next(x))
```

```python
print(next(x))
```

```python
print(next(x))
```

```python
print(next(x))
```

### Generator with loop

```python
def count_up(n):
    for i in range(1,n+1):
        yield i
```

### How Generator Works

----
📌 Steps:

1. Function is called -> No execution
2. Return generator object
3. next() reusmes execution
4. Stop at yield
5. Continues from same point

-----
#### Generator VS Normal Function

---
|Feature|Generator|Function|
|----|----|----|
|used|yield|return|
|Memory|Low|High|
|Execution|Lazy|Immediate|
|Values| One at a time| All at once|

#### Even Numbers Generator



```python
def even_numbers(n):
    for i in range(n):
        if i%2==0:
            yield i
```

```python
obj = even_numbers(10)
print(next(obj))
```

```python
print(next(obj))
```

### Generator with send()

```python
def gen():
    value = yield 1
    yield value
g = gen()
print(next(g))
```

```python
print(g.send(10))
```

### When use Generators:

* Large dataset
* Infinite sequence
* Streaming data
* Performance-critical apps

### When Not to use generators

* Need random access
* need all values repeatedly
