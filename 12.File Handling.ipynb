### Q. what is File Handling?
`File Handling` allows python program to read data files and write data to files permanently.

📌 Why we read files:
 * Stored data permanently.
 * Read logs, reports, CSV(Comma Seperated Value)
 * Process large data
 * Avoid manual input

### Types of Files
1. Text Files: [.txt,.csv,.log]
2. Binary Files: [.jpg,.pdf,.exe,.xlsx]

# ⤷ Steps to Read a file in Python

1. Open the file
2. Read the file content
3. Close the file

#### ⌯⌲ Opening a File
`file = open("fileName","mode")"`

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,"r")  # step 1

content = file.read()  # step 2
print(content)

file.close()   # step 3
```

## Modes of File
---
|Mode|Meaning|
|---|---|
|`r`|Read(default)|
|`w`|Write|
|`a`|append|
|`r+`| Read & write|

### Method 1: read()  --> Read Entire File



```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,"r")
content = file.read()
print(content)
file.close()
```

### Method 2: readline() --> Read one line at a time.


```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,"r")
content = file.readline()
content2 = file.readline()
content3 = file.readline()
print(content)
print(content2)
print(content3)
file.close()
```

### Method 3: readlines() -> Read all lines as list



```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,"r")
content = file.readlines()
print(content)
file.close()
```

### Syntax for Opening a file 

`open(fileName,"mode")`

#### Parameters 
|Parameters|Descrptions|
|-----|----|
|`fileName`|Path of the file including fileName|
|`mode`|Operations to perform on file|

### 3. File Modes
|Mode|Description|
|----|----|
|`r`|Read File(default)|
|`w`|Write file(Overwrite)|
|`a`|Append data|
|`x`|Create file if file exists it throw error|
|`rb`|Read binary|
|`wb`|write binary|

1. Read(n)
2. readline(n)

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,"r")
content = file.read(10)  # 10 chracters only
print(content)
file.close()
```

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,"r")
content = file.readline(10)  # 10 chracters
print(content)
file.close()
```

### Reading File Using Loop

Best Method for large files

#### Example

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,'r')

for line in file:
    print(line)
file.close()
```

### Using `with statement[Best Practice]`

with automatically closes the file.

syntax:

`with open(fileName,"mode") as variable:`

```python
with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as file:
    print(file.read())
```
