7. Take a number from the user and write its multiplication table into a file.


```python
num = int(input("Enter a number: "))

with open("multiplication_table.txt", "w") as file:
    for i in range(1, 11):
        file.write(f"{num} x {i} = {num * i}\n")

print("Multiplication table written to the file successfully.")
```

9. Create a list of fruits and write each fruit on a separate line in "fruits.txt".

```python
fruits = [
    "Apple",
    "Banana",
    "Orange",
    "Mango",
    "Grapes",
    "Pineapple",
    "Strawberry",
    "Watermelon",
    "Papaya",
    "Guava"
]

with open("fruits.txt", "w") as file:
    for fruit in fruits:
        file.write(fruit + "\n")
```

10. Take a student's name, roll number, marks, and grade and write them into a file.

```python
name = input("Enter student name: ")
roll_no = input("Enter roll number: ")
marks = input("Enter marks: ")
grade = input("Enter grade: ")

with open("student.txt", "w") as file:
    file.write(f"Student Name: {name}\n")
    file.write(f"Roll Number: {roll_no}\n")
    file.write(f"Marks: {marks}\n")
    file.write(f"Grade: {grade}\n")

print("Student details written to student.txt successfully.")
```
