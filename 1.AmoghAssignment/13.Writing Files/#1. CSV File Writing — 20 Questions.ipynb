8. Write a program to write a list of dictionaries into a CSV file.

12. Write a program to find the highest marks from a CSV file.

```python
import csv
```

```python
with open("students2.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)

    first_row = next(reader)
    highest = int(first_row[1])

    for row in reader:
        marks = int(row[1])
        if marks > highest:
            highest = marks

print("Highest Marks:", highest)            
```

13. Write a program to find the average marks from a CSV file.

```python
import csv
```

```python
total = 0
count = 0

with open("students2.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)

    for row in reader:
        total += int(row[1])
        count += 1

average = total / count

print("Average Marks:", average)
```

14. Write a program to search for a student by name in a CSV file.

```python
import csv
```

```python
name = input("Enter student name: ")

with open("students1.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        if row[0].lower() == name.lower():
            print("Student found:", row)
            break
    else:
        print("Student not found")
```

17. Write a program to separate students who passed and failed using a CSV file.

```python
import csv
```

```python

with open("students2.csv", "r") as file:
    data = csv.reader(file)

    with open("passed.csv", "w", newline="") as p, \
         open("failed.csv", "w", newline="") as f:

        passed = csv.writer(p)
        failed = csv.writer(f)

        header = next(data)
        passed.writerow(header)
        failed.writerow(header)

        for row in data:
            if int(row[1]) >= 40:
                passed.writerow(row)
            else:
                failed.writerow(row)
```

19. Write a program to copy the contents of one CSV file into another.

```python
import csv
```

```python
with open("employee.csv", "r") as file:
    reader = csv.reader(file)

    with open("employee1.csv", "w", newline="") as file:
        writer = csv.writer(file)

        for row in reader:
            writer.writerow(row)

print("File copied successfully")
```
