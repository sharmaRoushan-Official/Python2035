11. Take 10 numbers from the user and write them into a file.

```python
with open("numbers.txt", "w") as file:
    for i in range(10):
        num = input(f"Enter number {i + 1}: ")
        file.write(num + "\n")

print("10 numbers have been written to numbers.txt.")
```

12.  Write numbers from 1 to 100 into a file, but only numbers divisible by 5.

```python
with open("numbers.txt", "w") as file:
    for number in range(1, 101):
        if number % 5 == 0:
            file.write(f"{number}\n")
```

17. Write all prime numbers between 1 and 100 into a file.

```python
file = open("primes.txt", "w")

for n in range(2, 101):
    prime = True

    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    if prime:
        file.write(str(n) + "\n")

file.close()
```

19. Create a file containing multiplication tables from 1 to 10.

```python
with open("tables.txt", "w") as file:
    for i in range(1, 11):
        file.write(f"Multiplication Table of {i}\n")
        
        for j in range(1, 11):
            file.write(f"{i} x {j} = {i * j}\n")
        
        file.write("\n")

print("Multiplication tables from 1 to 10 written to tables.txt")
```

20. Write the first 20 Fibonacci numbers into a file.

```python
a = 0
b = 1

file = open("fibonacci.txt", "w")

for i in range(20):
    file.write(str(a) + "\n")
    c = a + b
    a = b
    b = c

file.close()

print("Fibonacci numbers written to file.")
```
