14. Write a program to create a PDF containing a simple invoice.

```python
pip install reportlab
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas

file = canvas.Canvas("invoice.pdf", pagesize=A4)

x = 50
y = 750

# Heading
file.setFont("Helvetica-Bold", 20)
file.drawString(x, y, "INVOICE")

# Invoice details
file.setFont("Helvetica", 12)
y = y - 40
file.drawString(x, y, "Invoice No: 101")
y = y - 20
file.drawString(x, y, "Date: 15-08-2026")

# Customer details
y = y - 40
file.drawString(x, y, "Customer: Rahul")
y = y - 20
file.drawString(x, y, "Address: Pune")

# Items
y = y - 40
file.drawString(x, y, "Item       Quantity       Price")
y = y - 25
file.drawString(x, y, "Laptop       1           50000")
y = y - 25
file.drawString(x, y, "Mouse        2             1000")


y = y - 40
file.drawString(x, y, "Total: Rs. 51000")


file.save()

print("Invoice PDF created successfully.")
```

15. Write a program to create a PDF containing a school timetable.

```python
pip install reportlab
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas

file = canvas.Canvas("timetable.pdf",pagesize=A4)

file.drawString(200, 800, "SCHOOL TIMETABLE")

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

subjects = [
    ["Math", "English", "Science"],
    ["English", "Math", "Computer"],
    ["Science", "English", "Math"],
    ["Computer", "Science", "English"],
    ["Math", "Sports", "Art"]
]

y = 750

for i in range(5):
    file.drawString(100, y, days[i])
    file.drawString(200, y, subjects[i][0])
    file.drawString(300, y, subjects[i][1])
    file.drawString(400, y, subjects[i][2])
    y -= 40

file.save()

print("PDF created successfully!")
```

17. Write a program to create a PDF containing a list of products and their prices.

```python
pip install reportlab
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas

file = canvas.Canvas("products.pdf",pagesize=A4)

products = [
    ("Pen", 10),
    ("Book", 50),
    ("Bag", 500),
    ("Pencil", 5)
]

y = 800

file.drawString(100, y, "Product List")
y -= 30

for product, price in products:
    file.drawString(100, y, product)
    file.drawString(250, y, "Rs. " + str(price))
    y -= 25

file.save()

print("PDF created successfully")
```

19. Write a program to create a multi-page PDF document.

```python
pip install reportlab 
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas

file = canvas.Canvas("document.pdf")

for page in range(1, 4):
    file.drawString(100, 750, f"This is page {page}")
    file.showPage()

file.save()

print("PDF created successfully!")
```
