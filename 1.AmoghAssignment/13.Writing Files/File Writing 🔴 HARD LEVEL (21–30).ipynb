27. Create a Bank Transaction file containing:

* Transaction ID
* Transaction Type
* Amount
* Balance

```python

n = int(input("Enter number of transactions: "))

with open("bank_transactions.txt", "w") as file:
    file.write("Transaction ID\tTransaction Type\tAmount\tBalance\n")

    for i in range(n):
        print(f"\nTransaction {i + 1}")

        transaction_id = input("Enter Transaction ID: ")
        transaction_type = input("Enter Transaction Type (Deposit/Withdrawal): ")
        amount = float(input("Enter Amount: "))
        balance = float(input("Enter Balance: "))

        file.write( f"{transaction_id}\t{transaction_type}\t{amount:.2f}\t{balance:.2f}\n"  )

print("\nBank transaction file created successfully!")
```

28. Create a Student Result File for 10 students containing:

* Name
* English Marks
* Math Marks
* Science Marks
* Total
* Percentage

```python
file = open("student_result.txt", "w")

for i in range(10):
    name = input("Enter student name: ")
    english = int(input("Enter English marks: "))
    math = int(input("Enter Math marks: "))
    science = int(input("Enter Science marks: "))

    total = english + math + science
    percentage = total / 3

    file.write(name + " ")
    file.write(str(english) + " ")
    file.write(str(math) + " ")
    file.write(str(science) + " ")
    file.write(str(total) + " ")
    file.write(str(percentage) + "%\n")

file.close()

print("Result file created successfully!")
```

29. Create a program to take employee information from the user and store it in a file.

The program should allow the user to add multiple employees until they choose to stop.

```python
file = open("employee.txt", "w")

while True:
    name = input("Enter employee name: ")
    age = input("Enter employee age: ")
    salary = input("Enter employee salary: ")

    file.write(name + " " + age + " " + salary + "\n")

    choice = input("Do you want to add another employee? (yes/no): ")

    if choice.lower() == "no":
        break

file.close()

print("Employee information saved successfully!")
```

30. Mini Project – Student Record Management System

Create a file-based Student Management System with the following menu:

1. Add Student
2. Add Multiple Students
3. Add Marks
4. Generate Result
5. Exit

Store all student information in a text file.

```python
def add_student():
    roll = input("Enter Roll Number: ")
    name = input("Enter Name: ")

    with open("students.txt4", "a") as file:
        file.write(roll + "," + name + ",-1\n")

    print("Student added successfully!")


def add_multiple_students():
    n = int(input("How many students? "))

    with open("students.txt", "a") as file:
        for i in range(n):
            print("\nStudent", i + 1)

            roll = input("Enter Roll Number: ")
            name = input("Enter Name: ")

            file.write(roll + "," + name + ",-1\n")

    print("\nStudents added successfully!")


def add_marks():
    roll = input("Enter Roll Number: ")
    marks = float(input("Enter Marks: "))

    try:
        with open("students.txt", "r") as file:
            lines = file.readlines()

        found = False

        with open("students.txt", "w") as file:
            for line in lines:
                data = line.strip().split(",")

                if data[0] == roll:
                    file.write(data[0] + "," + data[1] + "," + str(marks) + "\n")
                    found = True
                else:
                    file.write(line)

        if found:
            print("Marks added successfully!")
        else:
            print("Student not found!")

    except FileNotFoundError:
        print("No student records found.")


def generate_result():
    try:
        with open("students.txt", "r") as file:
            lines = file.readlines()

        print("\n--------------------------------")
        print("       STUDENT RESULT")
        print("--------------------------------")

        for line in lines:
            data = line.strip().split(",")

            roll = data[0]
            name = data[1]
            marks = float(data[2])

            print("\nRoll Number :", roll)
            print("Name        :", name)

            if marks == -1:
                print("Marks       : Not Added")
                print("Result      : Pending")
            else:
                print("Marks       :", marks)

                if marks >= 40:
                    print("Result      : PASS")
                else:
                    print("Result      : FAIL")

            print("--------------------------------")

    except FileNotFoundError:
        print("No student records found.")


# Main Program

while True:
    print("\n====================================")
    print(" STUDENT RECORD MANAGEMENT SYSTEM")
    print("====================================")
    print("1. Add Student")
    print("2. Add Multiple Students")
    print("3. Add Marks")
    print("4. Generate Result")
    print("5. Exit")
    print("====================================")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        add_multiple_students()

    elif choice == "3":
        add_marks()

    elif choice == "4":
        generate_result()

    elif choice == "5":
        print("Program terminated.")
        break

    else:
        print("Invalid choice! Please try again.")
```
