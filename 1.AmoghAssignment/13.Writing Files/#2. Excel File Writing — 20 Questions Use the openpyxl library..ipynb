2. Write a program to create an Excel file containing student names.

```python
pip install openpyxl
```

```python
from openpyxl import Workbook

workbook = Workbook()
sheet = workbook.active
sheet.title = "Students"

sheet["A1"] = "Student Name"

students = [
    "Aarav",
    "Priya",
    "Rahul",
    "Ananya",
    "Rohan"
]

for row, name in enumerate(students, start=2):
    sheet.cell(row=row, column=1, value=name)

workbook.save("students.xlsx")

print("Excel file created successfully: students.xlsx")
```

8. Write a program to write a list of dictionaries into an Excel worksheet.

```python
pip install openpyxl
```

```python
from openpyxl import openpyxl

wb = openpyxl.Workbook()
ws = wb.active

# Values
data = [
    {"Name": "John", "Age": 20},
    {"Name": "Alice", "Age": 22}
]

ws.append(["Name", "Age"])

for d in data:
    ws.append(list(d.values()))

wb.save("data.xlsx")
```

11. Write a program to find the highest marks in an Excel sheet.

```python
pip install openpyxl 
```

```python
from openpyxl import Workbook

wb = openpyxl.load_workbook("students_marks.xlsx")
sheet = wb.active

marks = []

for cell in sheet["B"]:
       if type(cell.value) == int:
           marks.append(cell.value)
print("Highest marks =", max(marks))
```

12. Write a program to calculate the average marks from an Excel sheet.

```python
pip install openpyxl
```

```python
from openpyxl import Workbook

wb = openpyxl.load_workbook("students_marks.xlsx")
sheet = wb.active

marks = []

for cell in sheet["B"]:
       if type(cell.value) == int:
           marks.append(cell.value)

average = sum(marks) / len(marks)

print("Average marks =", average)
```

13. Write a program to calculate the total salary of employees in Excel.

```python
pip install openpyxl
```

```python
from openpyxl import Workbook

wb = openpyxl.load_workbook("employee.xlsx")
sheet = wb.active

salary = []

for row in range(3, sheet.max_row + 1):
    salary.append(int(sheet.cell(row, 3).value))

print("Salary List =", salary)
print("Total Salary =", sum(salary))
```

19. Write a program to apply bold formatting to the header row of an Excel sheet.

```python
pip install openpyxl
```

```python
from openpyxl import Workbook
from copy import copy

wb = Workbook()
sheet = wb.active

sheet.append(["Name", "Age", "City"])

for cell in sheet[1]:
    font = copy(cell.font)
    font.bold = True
    cell.font = font

wb.save("output.xlsx")
```

20. Write a program to create a sales report in Excel containing product, quantity, price, and total amount.

```python
pip install openpyxl
```

```python
from openpyxl import Workbook

wb = Workbook()
sheet = wb.active

sheet["A1"] = "Product"
sheet["B1"] = "Quantity"
sheet["C1"] = "Price"
sheet["D1"] = "Total Amount"

sheet["A2"] = "Pen"
sheet["B2"] = 10
sheet["C2"] = 20
sheet["D2"] = sheet["B2"].value * sheet["C2"].value

sheet["A3"] = "Book"
sheet["B3"] = 5
sheet["C3"] = 100
sheet["D3"] = sheet["B3"].value * sheet["C3"].value

wb.save("sales_report.xlsx")
```
