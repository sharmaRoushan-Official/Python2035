11. Write a program to create a table of students and marks in DOCX.

```python
pip install python-docx
```

```python
from docx import Document

doc = Document()

table = doc.add_table(rows=1, cols=3)

table.rows[0].cells[0].text = "Name"
table.rows[0].cells[1].text = "Roll No"
table.rows[0].cells[2].text = "Marks"

data = [
    ("Rahul", "1", "85"),
    ("Priya", "2", "90"),
    ("Amit", "3", "78")
]

for name, roll, marks in data:
    row = table.add_row().cells
    row[0].text = name
    row[1].text = roll
    row[2].text = marks

doc.save("students.docx")
```

13. Write a program to create a shopping bill in DOCX.

```python
pip install python-docx
```

```python
from docx import Document

doc = Document()

doc.add_heading("Shopping Bill", level=1)

table = doc.add_table(rows=1, cols=4)

table.rows[0].cells[0].text = "Item"
table.rows[0].cells[1].text = "Quantity"
table.rows[0].cells[2].text = "Price"
table.rows[0].cells[3].text = "Total"

data = [
    ("Pen", "2", "10", "20"),
    ("Notebook", "3", "50", "150"),
    ("Pencil", "5", "5", "25")
]

for item, qty, price, total in data:
    row = table.add_row().cells
    row[0].text = item
    row[1].text = qty
    row[2].text = price
    row[3].text = total

doc.add_paragraph("Grand Total: 195")

doc.save("shopping_bill.docx")
```

14. Write a program to create an invoice in DOCX.

```python
pip install python-docx
```

```python
from docx import Document

doc = Document()

doc.add_heading("INVOICE", 0)

doc.add_paragraph("Invoice No: INV-001")
doc.add_paragraph("Date: 15 August 2026")

doc.add_paragraph("\nBill To:")
doc.add_paragraph("John Smith\nPune, India")

table = doc.add_table(rows=1, cols=4)
table.style = "Table Grid"

# Table headings
table.rows[0].cells[0].text = "Item"
table.rows[0].cells[1].text = "Quantity"
table.rows[0].cells[2].text = "Price"
table.rows[0].cells[3].text = "Total"

# Item
row = table.add_row().cells
row[0].text = "Laptop"
row[1].text = "1"
row[2].text = "50000"
row[3].text = "50000"

row = table.add_row().cells
row[0].text = "Mouse"
row[1].text = "2"
row[2].text = "500"
row[3].text = "1000"

doc.add_paragraph("\nGrand Total: ₹51,000")

doc.save("invoice.docx")

print("Invoice created successfully!")
```

15. Write a program to create a school timetable in DOCX.

```python
pip install python-docx
```

```python
from docx import Document

doc = Document()

doc.add_heading("School Timetable", 0)

table = doc.add_table(rows=4, cols=4)

data = [
    ["Day", "9:00-10:00", "10:00-11:00", "11:00-12:00"],
    ["Monday", "Maths", "English", "Science"],
    ["Tuesday", "English", "Maths", "Computer"],
    ["Wednesday", "Science", "English", "Maths"]
]

for i in range(4):
    for j in range(4):
        table.cell(i, j).text = data[i][j]

doc.save("school_timetable.docx")

print("Timetable created successfully!")
```

18. Write a program to create a DOCX document with a numbered list.

20. Write a program to create a formatted sales report containing product, quantity, price, and total amount.

```python
pip install python-docx
```

```python
from docx import Document

doc = Document()

product = "Pen"
quantity = 10
price = 20
total = quantity * price

doc.add_heading("Sales Report", 1)

table = doc.add_table(rows=2, cols=4)
table.style = "Table Grid"

table.cell(0, 0).text = "Product"
table.cell(0, 1).text = "Quantity"
table.cell(0, 2).text = "Price"
table.cell(0, 3).text = "Total Amount"

table.cell(1, 0).text = product
table.cell(1, 1).text = str(quantity)
table.cell(1, 2).text = str(price)
table.cell(1, 3).text = str(total)

doc.save("sales_report.docx")

print("Report created")
```
