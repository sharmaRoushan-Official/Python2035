21. Read a file and find the frequency of each word.

```python
with open(r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt",'r') as file:
    contents = file.read().lower()

    words = contents.split()

    frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("Word Frequencies:")
for word, count in frequency.items():
    print(f"{word} : {count}")
```

22.  Read a file and display the top 10 most frequently used words.

```python
fileName = r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt"

file = open(fileName,"r")

contents = file.read().lower()

file.close()

words = contents.split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print("Top 10 Most Frequent Words:")

count = 0
while freq and count < 10:
    max_word = ""
    max_freq = 0

    for word in freq:
        if freq[word] > max_freq:
            max_freq = freq[word]
            max_word = word

    print(max_word, ":", max_freq)
    del freq[max_word]
    count += 1
```
