11. Count the number of vowels present in a text file.

```python
fileName = r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt"

count = 0

file = open(fileName,"r")

for line in file:
    for ch in line:
        if ch in "aeiouAEIOU":
            count += 1
file.close() 

print("Number of vowels:", count)
```

12. Count the number of uppercase and lowercase letters in a file.

```python
fileName = r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt"

file = open(fileName,"r")

upper = 0
lower = 0

for ch in file.read():
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1


print("Uppercase letters:", upper)
print("Lowercase letters:", lower)


```

13. Find the longest word present in a file.

```python
fileName = r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt"

file = open(fileName,"r")

longest = ""

for word in file.read().split():
    if len(word) > len(longest):
        longest = word
        
file.close()

print("Longest word:", longest)
```

15. Count how many times a specific word appears in a file.

```python
fileName = r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt"

word = input("Enter word to search: ")

file = open(fileName, "r")

count = 0

for line in file:
    words = line.split()
    for w in words:
        if w == word:
            count += 1

file.close()

print("The word appears", count, "times")
```

17. Read a CSV file and calculate the average of a numeric column.

```python
import csv
```

```python
with open(r"C:\Users\Amogh\Downloads\data.csv","r") as file:
    
    reader = csv.DictReader(file)

    total = 0
    count = 0

    for row in reader:
        total += int(row["CO2"])
        count += 1

        average = total / count

print("Average:", average)    
```
