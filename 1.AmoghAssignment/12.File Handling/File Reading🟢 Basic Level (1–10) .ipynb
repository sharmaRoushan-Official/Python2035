8. Read a file and print only the even-numbered lines.

```python
with open(r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt",'r') as file:
    for line_number, line in enumerate(file, start=1):

         if line_number % 2 == 0:
            print(line, end="")
```

9. Read a file and print only the odd-numbered lines.

```python
fileName= r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt"

file = open(fileName,"r")

lines = file.readlines()

for i in range(0, len(lines), 2):
    print(lines[i], end="")

file.close()
```

10. Read a file and display each line with its line number.

```python
with open(r"C:\\Users\\Amogh\\OneDrive\\Desktop\\Softdel JD.txt",'r') as file:
    for line_number, line in enumerate(file, start=1):
        print(f"{line_number}: {line.rstrip()}")
```
