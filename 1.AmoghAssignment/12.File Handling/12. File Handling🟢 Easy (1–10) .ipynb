4. Count the total number of lines in a file.

```python
fileName = r"D:\\Amogh 2\\Amogh.txt"

file = open(fileName,"r")

count = 0

for line in file:
    count += 1

file.close()

print("Total number of lines:", count)
```

5. Count the total number of words in a file.

```python
fileName = r"D:\\Amogh 2\\Amogh.txt"

file = open(fileName,"r")

text = file.read()

word_count = len(text.split())

print("Total number of words =", word_count)

file.close()
```
