 11. Read a file and count how many times a given word appears.

```python
fileName = r"C:\Users\Amogh\OneDrive\Desktop\\Softdel JD.txt"

file = open(fileName,"r")

word = input("Enter the word to search: ")

count = 0

for line in file:
    words = line.split()
    for w in words:
        if w == word:
            count += 1

file.close()

print("The word appears", count, "times")
```

12. Read a file and print the longest line.

```python
with open(r"C:\Users\Amogh\OneDrive\Desktop\\Softdel JD.txt",'r') as file:
    longest_line = ""

    for line in file:
        if len(line) > len(longest_line):
            longest_line = line

print("Longest line in the file:")
print(longest_line)
print("Length:", len(longest_line))
```

13. Read a file and print the shortest line.

```python
fileName = r"D:\\AP\\Amogh.txt"

file = open(fileName,"r")

lines = file.readlines()

shortest = min(lines, key=len)

print("Shortest line:")
print(shortest)

file.close()
```

18. Read a file and count the number of vowels.

```python
fileName = r"D:\AP\Amogh.txt"

file = open(fileName, "r")

count = 0

for ch in file.read():
    if ch in "aeiouAEIOU":
        count += 1
        
file.close()

print("Number of vowels:", count)
```

19. Read a file and count the number of digits.

```python
fileName = r"D:\AP\Amogh.txt"

file = open(fileName, "r")

count = 0

for ch in file.read():
    if ch.isdigit():
        count += 1

file.close()

print("Number of digits:", count)
```

20. Read a file and print all unique words.

```python
fileName = r"D:\AP\Amogh.txt"

file = open(fileName, "r")

words = file.read().split()
unique = set(words)

print("Unique words")
for word in unique:
    print(word)

file.close()    
```
