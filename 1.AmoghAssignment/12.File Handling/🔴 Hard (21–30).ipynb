21. Compare two text files and display the lines that are different.

```python
with open(r"D:\\AP\\Amogh.txt",'r') as f1, open(r"D:\\AP\\comp.txt",'r') as f2:
    lines1 = f1.readlines()
    lines2 = f2.readlines()

    max_lines = max(len(lines1), len(lines2))

    for i in range(max_lines):
        line1 = lines1[i].rstrip() if i < len(lines1) else "<No Line>"
        line2 = lines2[i].rstrip() if i < len(lines2) else "<No Line>"

    if line1 != line2:
        print(f"Line {i + 1}:")
        print(f"  File1: {line1}")
        print(f"  File2: {line2}")
        print()
```

22.  Read a file and print duplicate words along with their occurrence count.

```python
fileName = r"D:\\AP\\Amogh.txt"

file = open(fileName,"r")

content = file.read()
file.close()

words = content.split()

count = {}

for word in words:
    if word in count:
        count[word] = count[word] + 1
    else:
        count[word] = 1

print("Duplicate words and their occurrence count:")
for word in count:
    if count[word] > 1:
        print(word, ":", count[word])
```

23. Find the most frequently occurring word in a file.

```python
fileName = r"D:\\AP\\Amogh.txt"

file = open(fileName,"r")

content = file.read()
file.close()

words = content.split()

count = {}

for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1

max_count = 0
max_word = ""

for word in count:
    if count[word] > max_count:
        max_count = count[word]
        max_word = word

print("Most frequent word:", max_word)
print("Occurrence count:", max_count)
```

28. Read a file and create a frequency dictionary of all words.

29. Read a text file and determine whether it is a palindrome after removing spaces, punctuation, and case differences.

```python
fileName = r"D:\\AP\\Amogh.txt"

file = open(fileName, "r")

content = file.read()

file.close()

content = ''.join(ch.lower() for ch in content if ch.isalnum())

if content == content[::-1]:
    print("The file is a palindrome")
else:
    print("The file is not a palindrome")
```

30. Create a menu-driven program with the following options:



Read entire file

Count lines

Count words

Search for a word

Display the longest line

Exit

```python
fileName = r"D:\\AP\\Amogh.txt"

file = open(fileName, "r")

data = file.read()

file.close()

while True:
    print("\n1.Read file")
    print("2.Count lines")
    print("3.Count words")
    print("4.Search word")
    print("5.Longest line")
    print("6.Exit")

    ch = int(input("Enter choice: "))

    if ch == 1:
        print(data)

    elif ch == 2:
        print("Lines:", len(data.splitlines()))

    elif ch == 3:
        print("Words:", len(data.split()))

    elif ch == 4:
        w = input("Enter word: ")
        print("Found" if w in data else "Not found")

    elif ch == 5:
        print(max(data.splitlines(), key=len))

    elif ch == 6:
        break

    else:
        print("Wrong choice")
        
```
