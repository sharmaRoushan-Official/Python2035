45. Write a recursive function to find the sum of all elements in a list.

```python
def list_sum(lst):
    if len(lst) == 0:
        return 0
    return lst[0] + list_sum(lst[1:])
```

```python
print(list_sum([1, 2, 3, 4]))  
```

47. Write a function that reads a file and returns the number of lines.

```python
def count_lines(filename):
    with open(filename, "r") as file:
        return len(file.readlines())
```

```python
print(count_lines("example.txt"))
```

48. Write a function that returns a list of even numbers using list comprehension.

```python
def even_numbers(n):
    return [x for x in range(n + 1) if x % 2 == 0]
```

```python
print(even_numbers(10))
```

49. Write a function that takes a list of strings and returns a new list with all strings in uppercase.

```python
def to_uppercase(lst):
    return [s.upper() for s in lst]
```

```python
print(to_uppercase(["hello", "world"]))
```
