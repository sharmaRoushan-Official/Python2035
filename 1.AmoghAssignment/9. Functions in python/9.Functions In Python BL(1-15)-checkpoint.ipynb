6. Write a function that takes a list of numbers and returns the largest number.

```python
def largest_number(numbers):
    if not numbers:
        raise ValueError("The list is empty.")
    return max(numbers)
```

```python
nums = [3, 7, 2, 9, 5]
print(largest_number(nums))  
```

7. Write a function that calculates the factorial of a number.

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```

```python
print(factorial(5))  
```

9. Write a function to count the number of vowels in a string.

```python
def count_vowels(s):
    vowels = "aeiouAEIOU"
    count = 0
    for char in s:
        if char in vowels:
            count += 1
    return count
```

```python
print(count_vowels("Hello, World!"))  # Output: 3
print(count_vowels("Python"))         # Output: 1
```

11. Write a function that takes two strings and checks if they are anagrams.

```python
def are_anagrams(str1, str2):
    return sorted(str1) == sorted(str2)
```

```python
print(are_anagrams("listen", "silent"))  
print(are_anagrams("hello", "world")) 
```

13. Write a function to check if a number is prime.

```python
def prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True    
    
```

```python
print(prime(7))
print(prime(10))
```

14. Write a function that prints the Fibonacci sequence up to n terms. 

```python
def fibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        print(a)
        c = a + b
        a = b
        b = c
```

```python
fibonacci(5)
```
