16. Write a function that accepts a list of integers and returns the sum of even numbers. 

```python
def sum_of_evens(numbers):
    total = 0
    for n in numbers:
        if n % 2 == 0:
            total += n
    return total
```

```python
print(sum_of_evens([1, 2, 3, 4, 5, 6]))  
```

18. Write a function that takes a number and returns the sum of its digits.

```python
def sum_of_digits(n):
    n = abs(n)
    s = 0

    for digit in str(n):
        s += int(digit)

    return s
```

```python
print(sum_of_digits(1234))
print(sum_of_digits(90))
```

23. Write a function to find the second largest number in a list.

```python
def second_largest(nums):
    nums = list(set(nums))  # remove duplicates
    nums.sort()
    
    if len(nums) < 2:
        return None
    
    return nums[-2]
```

```python
print(second_largest([10, 20, 4, 45, 99]))  
print(second_largest([5, 5, 5]))  
```

24. Write a function that takes a string and counts uppercase and lowercase letters.

```python
def count_case(s):
    upper = 0
    lower = 0

    for ch in s:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1

    return upper, lower
```

```python
print(count_case("Hello World"))  
print(count_case("Python123"))   
```

26. Write a function that takes variable number of arguments and returns their average.

```python
def average(*args):
    if len(args) == 0:
        return 0  
    return sum(args) / len(args)
```

```python
print(average(10, 20, 30))   
print(average(5, 15))        
print(average())             
```

27. Write a function to remove punctuation from a string.

```python
def remove_punctuation(s):
    return "".join(ch for ch in s if ch.isalnum() or ch.isspace())
```

28. Write a function to calculate the compound interest.

```python
def compound_interest(p, r, t, n=1):
    return p * (1 + r / (n * 100)) ** (n * t) - p
```

```python
print(compound_interest(1000, 5, 2))  
```

29. Write a recursive function to calculate factorial.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
```

```python
print(factorial(4))
```

30. Write a recursive function to compute the nth Fibonacci number.

```python
def fibonacci(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
print(fibonacci(6))
```

31. Write a function to check if a given year is a leap year.

```python
def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    return False
```

```python
print(is_leap_year(2024))  
print(is_leap_year(1900))  
print(is_leap_year(2000))  
```

33.  Write a function that returns all prime numbers up to a given number n

```python
def primes_up_to(n):
    primes = []
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
    return primes
```

```python
print(primes_up_to(20))  
```

34.  Write a function that takes a list and returns a new list with only unique elements (preserving order).

```python
def unique_elements(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
```

```python
print(unique_elements([1, 2, 2, 3, 1, 4]))
```
