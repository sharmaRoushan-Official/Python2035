1. Welcome Decorator
Create a decorator that prints "Welcome" before executing the function and "Thank You" after the function finishes.

```python
def welcome_decorator(func):
    def wrapper():
        print("Welcome")
        func()
        print("Thank You")
    return wrapper

@welcome_decorator
def greet():
    print("Hello")
greet()    
```

2. Execution Time Decorator
Create a decorator that calculates and prints the execution time of a function.

```python
import time

def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print("Execution time:", end - start, "seconds")
    return wrapper

@timer
def hello():
    print("Hello World")

hello()
```

3. Login Required Decorator
Create a decorator that allows a function to execute only if logged_in = True; otherwise print "Please login first."

```python
def login_required(func):
    def wrapper():
        if logged_in:
            return func()
        else:
            print("Please login first.")
    return wrapper


logged_in = True

@login_required
def dashboard():
    print("Welcome to your dashboard!")


dashboard()
```

4. Function Call Counter
Create a decorator that counts and displays how many times a function has been called.

```python
def counter(func):
    count = [0]

    def wrapper():
        count[0] += 1
        print("Called", count[0], "times")
        func()

    return wrapper


@counter
def hello():
    print("Hello")


hello()
hello()
hello()

        
```

5. Uppercase Output Decorator
Create a decorator that converts the string returned by a function into uppercase before displaying it.

```python
def uppercase_output(func):
    def wrapper():
        result = func()
        return result.upper()
    return wrapper


@uppercase_output
def greet():
    return "hello, world!"


print(greet())
```

6. Repeat Function Decorator
Create a decorator that executes a function exactly 3 times whenever it is called.

```python
def repeat_three_times(func):
    def wrapper(*args, **kwargs):
        for _ in range(3):
            func(*args, **kwargs)
    return wrapper


@repeat_three_times
def greet():
    print("Hello!")


greet()
```

7. Argument Logger Decorator
Create a decorator that prints all the arguments passed to a function before executing it and then prints the returned result.

```python
def argument_logger(func):
    def wrapper(*args, **kwargs):
        print("Arguments:", args)
        print("Keyword arguments:", kwargs)

        result = func(*args, **kwargs)

        print("Returned result:", result)
        return result

    return wrapper


@argument_logger
def add(a, b):
    return a + b


add(5, 3)
```

8. Even Number Decorator
Create a decorator that checks whether the given number is even before executing the function. If the number is odd, print "Only even numbers are allowed."

```python
def even_only(func):
    def wrapper(num):
        if num % 2 == 0:
            return func(num)
        else:
            print("Only even numbers are allowed.")
    return wrapper


@even_only
def show_number(num):
    print(f"The number is {num}")


show_number(10)
show_number(7)
```

9. Password Protection Decorator
Create a decorator that asks the user for a password before executing the function. If the password is correct, execute the function; otherwise print "Access Denied."

```python
def password_protection(func):
    def wrapper():
        password = input("Enter Password: ")

        if password == "1234":
            func()
        else:
            print("Acess Denied. ")
    return wrapper

@password_protection
def secret_function():
    print("Access Granted! Welcome to the secret function.")

secret_function()
```

10. Exception Handling Decorator
Create a decorator that catches any exception raised by a function and prints the exception message instead of stopping the program.

```python
def handle_exceptions(func):
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except Exception as e:
            print("Exception:",e)
    return wrapper

@handle_exceptions
def divide(a, b):
    return a / b

print(divide(10, 2))
print(divide(10, 0))
```

11. Positive Number Decorator

```python
def positive_number(func):
    def wrapper(num):
        if num > 0:
            return func(num)
        else:
            print("Invalid Number.")
    return wrapper


@positive_number
def display(num):
    print("Number is:", num)


display(10)
display(-5)
```

12. Authorization Decorator
Create a decorator that allows only the user with role = "admin" to execute the function; otherwise print "Unauthorized Access."

```python
def admin_only(func):
    def wrapper(role):
        if role == "admin":
            func()
        else:
            print("Unauthorized Access.")
    return wrapper


@admin_only
def welcome():
    print("Welcome Admin!")


welcome("admin")
welcome("user")
```

13. Reverse String Decorator
Create a decorator that reverses the string returned by a function.

```python
def reverse_string(func):
    def wrapper():
        result = func()
        return result[::-1]
    return wrapper


@reverse_string
def get_string():
    return "Hello World"


print(get_string())
```

16. Input Validation Decorator
Create a decorator that checks if all arguments passed to a function are integers. If not, print "Invalid Input."

```python
def check(func):
    def wrapper(*args):
        for i in args:
            if not isinstance(i, int):
                print("Invalid Input.")
                return
        func(*args)
    return wrapper

@check
def add(a, b):
    print(a + b)

add(10, 20)
add(10, "20")
```

17. Minimum Age Decorator
Create a decorator that allows a function to execute only if age is 18 or above.

```python
def minimum_age(func):
    def wrapper(age, *args, **kwargs):
        if age >= 18:
            return func(age, *args, **kwargs)
        else:
            return "Access denied: You must be 18 or older."
    return wrapper


@minimum_age
def enter_club(age):
    return "Welcome to the club!"


print(enter_club(20))
print(enter_club(16))
```

18. HTML Tag Decorator
Create a decorator that wraps the returned string inside <h1>...</h1> HTML tags.

```python
def html_tag_decorator(func):
    def wrapper():
        return f"<h1>{func()}</h1>"
    return wrapper


@html_tag_decorator
def get_text():
    return "Hello World"


print(get_text())
```

19. Debug Decorator
Create a decorator that prints the function name, arguments, and returned value whenever the function is called.

```python
def debug(func):
    def wrapper(*args):
        print("Function:", func.__name__)
        print("Arguments:", args)
        result = func(*args)
        print("Returned:", result)
        return result
    return wrapper

@debug
def add(a, b):
    return a + b

add(3, 5)
```

20. Retry Decorator
Create a decorator that retries a function up to 3 times if it raises an exception before finally displaying the error.

```python
def retry(func):
    def wrapper():
        for i in range(3):
            try:
                func()
                return
            except Exception as e:
                if i == 2:
                    print("Error:", e)
    return wrapper


@retry
def test():
    print("Running...")
    raise Exception("Something went wrong")


test()
```
