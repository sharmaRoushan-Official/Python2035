11. Create a class Account with private attributes __balance. Provide methods deposit() and withdraw() to modify it safely.

```python
class Account:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive.")


    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount <= self.__balance:
            self.__balance = amount
            print(f"Withdrawn: {amount}")
        else:
            print("Insufficent Balance")

    def get_balance(self):
        return self.__balance 
            
```

```python
account = Account(1000)

account.deposit(500)
account.withdraw(300)

print("Current balance:", account.get_balance())
```

12. Write a class with a private method __calculate_salary() and a public method get_salary() that internally calls the private method.

```python
class Employee:
    def __init__(self, basic_salary, bonus):
        self.basic_salary = basic_salary
        self.bonus = bonus

    # Private method
    def __calculate_salary(self):
        return self.basic_salary + self.bonus

    # Public method
    def get_salary(self):
        return self.__calculate_salary()


# Example usage
employee = Employee(50000, 5000)

print("Total Salary:", employee.get_salary())
```

13. Create a class Student with private attributes __marks. Write a method to set marks (with validation: marks must be 0–100).

```python
class Student:
    def __init__(self):
        self.__marks = 0  # Private attribute

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
            print("Marks set successfully.")
        else:
            print("Invalid marks! Marks must be between 0 and 100.")

    def get_marks(self):
        return self.__marks


# Example usage
s = Student()

s.set_marks(85)    # Valid
print("Marks:", s.get_marks())

s.set_marks(120)   # Invalid
print("Marks:", s.get_marks())
```

14. Create a class Product with a private attribute __price. Use getter and setter methods to access and update price.

```python
class Product:
    def __init__(self):
        self._price = 0

    def set_price(self, price):
        self.__price = price

    def get_price(self):
        return self.__price

p = Product()

p.set_price(1500)
print("Price:", p.get_price())

p.set_price(2000)
print("Updated Price:", p.get_price())        
```

15. Create a class with a protected attribute _data and a child class that modifies and prints the protected attribute.

```python
class Parent:
    def __init__(self):
        self._data = "Hello"   # Protected attribute


class Child(Parent):
    def modify_data(self):
        self._data = "Welcome"

    def display_data(self):
        print("Protected Data:", self._data)


# Example usage
c = Child()
c.display_data()     # Before modification
c.modify_data()
c.display_data()     # After modification
```
