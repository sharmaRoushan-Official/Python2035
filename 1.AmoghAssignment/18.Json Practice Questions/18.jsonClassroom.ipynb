9. Given a JSON object with nested address, write code to access the city.

```python
import json

data = {
    "address": {
        "street": "123 Main St",
        "city": "Pune",
        "state": "Maharashtra"
    }
}

json_string = json.dumps(data)


obj = json.loads(json_string)


city = obj["address"]["city"]

print(city)  
```

17. Write code to update a student's marks.

```python
students = {
    101: {"name": "Aarav", "marks": 75},
    102: {"name": "Priya", "marks": 82},
}

def update_marks(student_id, new_marks):
    if student_id not in students:
        print("Student not found.")
        return

    if not 0 <= new_marks <= 100:
        print("Marks must be between 0 and 100.")
        return

    students[student_id]["marks"] = new_marks
    print(f"Marks updated for {students[student_id]['name']}.")
          
update_marks(101, 88)

print(students[101])
```

19. Write code to find the student with the highest marks.

```python
students = {
    "Rahul": 85,
    "Priya": 92,
    "Amit": 78,
    "Sneha": 95,
    "Rohan": 88
}

highest_student = max(students, key=students.get)

print("Student with highest marks:", highest_student)
print("Marks:", students[highest_student])
```

20. Write code to find the student with the lowest marks.

```python
students = {
    "Rahul": 85,
    "Priya": 92,
    "Amit": 78,
    "Sneha": 95,
    "Rohan": 88
}

lower_student = min(students, key=students.get)

print("Student with lower marks:", lower_student)
print("Marks:", students[lower_student])
```

22. Write code to calculate the total price of all products in the shopping cart.

```python
cart = [
    {"name": "Laptop", "price": 1000, "quantity": 1},
    {"name": "Mouse", "price": 25, "quantity": 2},
    {"name": "Keyboard", "price": 50, "quantity": 1}
]

total = sum(product["price"] * product["quantity"] for product in cart)

print(f"Total: ${total:.2f}")
```

29. Write code to sort students according to their marks.

```python
students = [
    {"name": "Rahul", "marks": 75},
    {"name": "Priya", "marks": 92},
    {"name": "Amit", "marks": 85}
]

students.sort(key=lambda x: x["marks"], reverse=True)

for student in students:
    print(student["name"], student["marks"])
```
