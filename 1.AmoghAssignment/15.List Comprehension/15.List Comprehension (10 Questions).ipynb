12. Create a list of even numbers from 1 to 50 using list comprehension.

```python
even_numbers = [num for num in range(1, 51) if num % 2 == 0]
print(even_numbers)
```

13. Convert the list of strings to uppercase using list comprehension:

```python
words = ['cat', 'dog', 'lion']
[i.upper() for i in words]
```

14. Get a list of numbers divisible by both 3 and 5 from 1 to 100.

```python
numbers = [i for i in range(1, 101) if i % 3 == 0 and i % 5 == 0]
print(numbers)
```

15. Extract all vowels from the string using list comprehension:
    text = "Comprehension"

```python
text = "Comprehension"
vowels = [char for char in text if char.lower() in "aeiou"]
print(vowels)
```

16. Given a matrix:
    matrix = [[1,2,3],[4,5,6],[7,8,9]]
    Use list comprehension to flatten it into a single list.

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]

flattenList = []
for i in matrix:
    for j in i:
        #print(j)
        flattenList.append(j)
print(flattenList)        
```

17. Given a list of words, extract words that contain the letter ‘e’:
    words = ['tree', 'book', 'python', 'sky']

```python
words = ['tree', 'book', 'python', 'sky']

result = [word for word in words if 'e' in word]

print(result)
```

18. Create a list of all 3-digit numbers that are palindromes using list comprehension.

```python
palindromes = [num for num in range(100, 1000) if str(num) == str(num)[::-1]]

print(palindromes)
```

20. Using nested list comprehension, generate a 3x3 identity matrix.

```python
identity_matrix = [[1 if i == j else 0 for j in range(3)] for i in range(3)]
print(identity_matrix)
```
