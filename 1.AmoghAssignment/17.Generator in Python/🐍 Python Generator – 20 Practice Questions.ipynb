1️⃣ Write a generator function that yields numbers from 1 to 10.

```python
def numbers():
    for i in range(1, 11):
        yield i
```

```python
for number in numbers():
    print(number)
```

2️⃣ Write a generator that yields all even numbers from 1 to N.

```python
def even_numbers(N):
    for num in range(2, N + 1, 2):
        yield num

# Example
for n in even_numbers(10):
    print(n)
```

6️⃣ Write a generator that yields the Fibonacci sequence up to N terms.

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

for num in fibonacci(10):
    print(num)
```

7️⃣ Write a generator that yields the factorials of numbers from 1 to N.

```python
def factorials(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
        yield fact



for value in factorials(5):
    print(value)
```

8️⃣ Write a generator that yields all prime numbers between 1 and N.

```python
def prime_generator(N):
    for num in range(2, N + 1):
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                break
        else:
            yield num


N = int(input("Enter N: "))

for prime in prime_generator(N):
    print(prime)
```

9️⃣ Write a generator that yields each character of a given string one by one.

```python
def char_generator(string):
    for char in string:
        yield char


for ch in char_generator("Hello"):
    print(ch)
```

🔟 Write a generator that yields each word from a given sentence.

```python
def word_generator(sentence):
    for word in sentence.split():
        yield word


sentence = "Python is easy to learn"

for word in word_generator(sentence):
    print(word)
```

1️⃣4️⃣ Write a generator that yields the running sum of numbers in a list.
Example:
Input: [1, 2, 3, 4]
Output: 1, 3, 6, 10

```python
def running_sum(numbers):
    total = 0
    for num in numbers:
        total += num
        yield total



numbers = [1, 2, 3, 4]

for value in running_sum(numbers):
    print(value)
```

1️⃣6️⃣ Create an infinite generator that keeps yielding natural numbers starting from 1.

```python
def natural_numbers():
    n = 1
    while True:
        yield n
        n += 1

gen = natural_numbers()

print(next(gen))  
print(next(gen))  
print(next(gen))  
```

1️⃣7️⃣ Write a generator that yields all palindrome numbers between 1 and N.

```python
def palindrome_numbers(n):
    for i in range(1, n + 1):
        if str(i) == str(i)[::-1]:
            yield i


N = 200
for num in palindrome_numbers(N):
    print(num)
```

1️⃣8️⃣ Write a generator that yields the binary representation of numbers from 1 to N.

```python
def binary_generator(N):
    for i in range(1, N + 1):
        yield bin(i)[2:]
```

```python
for binary in binary_generator(5):
    print(binary)
```

1️⃣9️⃣ Write a generator that yields numbers from 1 to N:
✔ Divisible by 3 → "Fizz"
✔ Divisible by 5 → "Buzz"
✔ Divisible by both → "FizzBuzz"

```python
def fizz_buzz_generator(n):
    for i in range(1, n + 1):
        if i % 15 == 0:
            yield "FizzBuzz"
        elif i % 3 == 0:
            yield "Fizz"
        elif i % 5 == 0:
            yield "Buzz"
        else:
            yield i



for value in fizz_buzz_generator(15):
    print(value)
```

2️⃣0️⃣ Write a generator that accepts a list and yields only unique elements while preserving the original order.
Example:
Input: [1, 2, 2, 3, 4, 4, 5]
Output: 1, 2, 3, 4, 5

```python
def unique_elements(items):
    seen = set()
    for item in items:
        if item not in seen:
            seen.add(item)
            yield item


nums = [1, 2, 2, 3, 4, 4, 5]

print(*unique_elements(nums))
```

🎯 Bonus Challenge:
Write your own version of Python's enumerate() using a generator that yields (index, value) pairs for any iterable.

```python
def my_enumerate(iterable, start=0):
    index = start
    for value in iterable:
        yield index, value
        index += 1



for i, value in my_enumerate(["apple", "banana", "cherry"]):
    print(i, value)
```
