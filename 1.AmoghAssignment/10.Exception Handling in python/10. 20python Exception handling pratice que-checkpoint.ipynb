### *1. Basic Syntax*

Write a Python program that takes a number from the user and prints its square. Use try and except to handle invalid (non-numeric) input.

### *9. Try With Return Statement*

Write a function using try-except-finally where the try returns a value but the finally also contains a return. Observe the output.

```python
def test():
    try:
        print("Inside try block")
        return 10

    except Exception:
        print("Inside except block")
        return 20

    finally:
        print("Inside finally block")
        return 30

result = test()
print("Returned value:", result)
```

### *10. Custom Exception*

Create a custom exception InvalidMarkError and raise it when marks > 100 or < 0. Handle it using try-except.

```python
# Custom exception class
class InvalidMarkError(Exception):
    pass

try:
    marks = int(input("Enter marks: "))

    if marks < 0 or marks > 100:
        raise InvalidMarkError("Marks should be between 0 and 100.")

    print("Marks entered:", marks)

except InvalidMarkError as e:
    print("Error:", e)

except ValueError:
    print("Error: Please enter a valid integer.")
```

## *15. Try-Except Inside a Function*

Create a function that calculates the square root of a number. Handle ValueError if input < 0.

```python
def sqrt_number():
    try:
        num = float(input("Enter a number: "))
        
        if num < 0:
            raise ValueError("Negative number not allowed.")
        
        result = num ** 0.5
        print("Square root:", result)
    
    except ValueError:
        print("Invalid input or negative number not allowed.")

sqrt_number()
```

### *16. Reading From a Possibly Corrupt File*

Write a program that tries to read and convert each line of a file into an integer. Handle:

* FileNotFoundError
* ValueError (line cannot convert)

```python
try:
    file = open("D:\\Amogh\\Password 123.txt", 'r')

    for line in file:
        try:
            num = int(line.strip())
            print("Valid integer:", num)
        except ValueError:
            print("Invalid line (not an integer):", line.strip())

    file.close()

except FileNotFoundError:
    print("File not found.")
```
