# Class Method (`@classmethod`)

### Definition
A **class method** is a method that works with the **class** instead of a specific object (instance). It is defined using the `@classmethod` decorator and takes **`cls`** as its first parameter.

### Syntax

```python
class ClassName:

    @classmethod
    def methodName(cls):
        # class-level code
        pass
```

### Key Points

- A class method is created using the **`@classmethod`** decorator.
- The first parameter is **`cls`**, which refers to the **class** itself.
- It can access and modify **class variables**.
- It **cannot directly access instance variables** because it does not receive `self`.
- It can be called using **both the class name and an object**.
- Mainly used for working with **class-level data** shared by all objects.
- It is commonly used to create **factory methods** and perform operations related to the class.

### Example

```python
class Student:
    schoolName = "Podar International School"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def schoolDetails(cls):
        print("School Name:", cls.schoolName)

obj = Student("Avnish", 13)

obj.schoolDetails()        # Calling using object
Student.schoolDetails()    # Calling using class
```

### Output

```
School Name: Podar International School
School Name: Podar International School
```

### Remember

- **`self` → Refers to the current object (instance).**
- **`cls` → Refers to the class itself.**
- **Instance Method → Works with instance variables.**
- **Class Method → Works with class variables.**

````python
class Student:
    schoolName = "Podar International School"  # access directly using object not in method 
    def __init__(self,name,age):
        self.name = name
        self.age = age

    # instance method
    def studentDetails(self):  # method
        print("Student Name:",self.name)
        print("Student Age:",self.age)


    # class method 
    @classmethod
    def schoolDetails(cls):
        print(f"School Name:{cls.schoolName}")


    @staticmethod
    def avnishEmail():  #static Method 
        print(f"Email of Avnish:avnish@gmail.com")

obj = Student("Avnish",13)
obj.studentDetails()
obj.schoolDetails()
obj.avnishEmail()
````

### **Static Method**

A **static method** is a method that **does not use `self` (instance)** or **`cls` (class)**. It behaves like a normal function but is defined inside a class using the `@staticmethod` decorator.

**Syntax:**
```python
@staticmethod
def method_name():
    # code
```

**Example:**
```python
class Student:

    @staticmethod
    def avnishEmail():
        print("Email of Avnish: avnish@gmail.com")
```

**Calling a Static Method:**
```python
Student.avnishEmail()   # Using class name
```

or

```python
obj = Student("Avnish", 13)
obj.avnishEmail()       # Using object
```

**Output:**
```
Email of Avnish: avnish@gmail.com
```

### **Key Points**
- Uses the `@staticmethod` decorator.
- Does **not** take `self` or `cls` as the first parameter.
- Cannot directly access instance variables or class variables.
- Can be called using **either the class name or an object**.
- Used for **utility/helper functions** related to the class but independent of class and object data.

### 1. Single Inheritance

Meaning:

In `Single Inheritance`, a child class inherits from only one parent class.

📌 Structure:

``Parent ➽ Child``

### Why It is used
 * Code Reusebility
 * Child class can use prent class properties and methods

### Real-life Example

 * Parent: Vehicle
 * Child: Car

A Car is a Vehicle

````python
class Vehicle:  # parent
    def start(self):
        print("Vehicle Started")


class Car(Vehicle):  # child of Vehicle
    def drive(self):
        print("Car is driving")

# v = Vehicle()
# v.start()

c = Car()
c.start()
c.drive()
````

### 3. Multilevel Inheritance


Meaning

In `Multilevel Inheritance`, inheritance happens in levels.

A class inherits from a class, and another class inherits from that child class.


📌 Structure

```text
Grandparent ➽ Parent ➽ Child
```

#### Why it is used

* To represent hierarchy in real world.
* Step-by-step inheritance of features.

#### Real-life Example
* Grandparent- Animal
* Parent - Mammal
* Child - Dog

##### Dog is a Mammal, Mammal is an animal

Python Example

````python
class Animal:  # parent
    def eat(self):
        print("Eating......")

class Mammal(Animal):  # mammal is a child of Animal
    def walk(self):
        print("Walking.....")

class Dog(Mammal):  # Dog is a child of Mammal
    def bark(self):
        print("Barking.....")

d = Dog()
d.eat()
d.walk()
d.bark()
````

Note:
The child class gets access to all previous level methods.

````python
print(Dog.mro())
````

### 4. Hierarchical Inheritance

Meaning

In `Hierarchical Inheritance`, multiple child classes inherits from a single parent class.

📌 Structure

```text
         Parent
       /    |   \
parent1 parent2 parent3
```

### Why it is used

 * One parent has common functionality
 * Differnt children use and extend it differently.

### Real-life Example

Parent: Employee

Children:
 * Manager : managing Work
 * Developer : Developer code
 * Tester: Testing Code

All are employees but have different roles.


##### Python Example

````python
class Employee:  # parent
    def work(self):
        print("Employee is working.....")

class Manager(Employee):  # child of Employee
    def manage(self):
        print("Manager is managing.....")
        
class Developer(Employee):  # child of Employee
    def code(self):
        print("Developer is coding......")

class Tester(Employee):  # child of Employee
    def test(self):
        print("Tester is testing......")


m = Manager()
m.work() # work is the method of parent
m.manage() # manage is the method of manager

d = Developer()
d.work()  # work is the method of parent
d.code()  # code is the method of Developer
 
t = Tester()
t.work()  # work is the method of parent
t.test() # test is the method of tester
````

### Note:

This is very common in software systems (Banking, college systems,  Employee Management)


````python
print(Manager.mro())
````

````python
print(Developer.mro())
````

````python
print(Tester.mro())
````

### Hybrid Inheritance

Meaning

Hybrid Inheritance is a combination of two or more inheritance types.

It may inclues:

* Multiple + Multilevel
* Hierarchical + Multiple

📌 Structure

```text
        A
      /  \
     B    C
      \  /
       D
```

Here:

* B and C inhertis from A [Hierarchical]
* D inhertis B and C [Multiple]

### Why It is used

* When a system requires complex relationships
* Common in large frameworks and enterprises projects

### Python Example

````python
class A:  # parent
    def showA(self):
        print("Class A")

class B(A): # child of A
    def showB(self):
        print("Class B")

class C(A): # child of A
    def showC(self):
        print("Class C")

class D(B,C):  # D is a child of B, C
    def showD(self):
        print("Class D")


obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
````

Notes:

Hybrid Inheritance is powerful but can cuase confusion.

* Method conflicts
* ambiguity
That's why Python uses MRO(Method Resolution Order)

````python
print(D.mro())
````

### 2.Encapsulation

Meaning

In encapsulation, we contorl access to class varaibles/methods using access modifiers. 
Python doest not enforce them strictly like java, but it follows naming conventions.

Python has 3 type of access modifiers:

1. Public
2. Protected
3. Private

## 1. Public Members/Variables

* Accessible everywhere
* No underscore prefix

````python
class Student:
    def __init__(self,name):
        self.name = name  # public variabe

    def display(self):  # public method
        print("Student Name:",self.name)

s = Student("Shyam")
print(s.name)  # public member 
s.display()  # public method
````

### 2. Protected Members(_)

* Should be accessed only inside class and child class
* Defined using single underscore `_varaible`

Note: Still accessible outside, but considered not recommended.

#### Example

````python
class Employee:
    def __init__(self,salary):
        self._salary = salary  # protected variable

class Manager(Employee):
    def show_salary(self):
        print("Salary:",self._salary)


m = Manager(50000)
m.show_salary()  # salary calling using method show_salary()

print(m._salary)  ## possible, but not recommended
````

## 3. Private Members(__)

* Accessble only inside the class.
* Defined using double underscore `__variable`
* Python applies name mangling(internally changes name)

#### Example

````python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance # private variable

    def show_balance(self):  # method of class
        print("Balance:",self.__balance)

acc = BankAccount(50000)
acc.show_balance()

# print(acc.__balance)  # Error (not accessible outside the class)
````

### Accessing Private Variable

Even private memebers can be accessed like this (not recommended):

````python
print(acc._BankAccount__balance)
````

### Summary Table

|Modifier Type|Symbol|Access Scope|
|----|----|----|
|Public| name| Everywhere|
|Protected| `_name`|Class+Child class|
|Private|`__name`|Only inside class|

### Encapsulation using Getter & Setter Methods

````python
class Student:
    def __init__(self):
        self.__marks = 0  ## private members

    # set marks
    def set_marks(self,marks):
        if marks>=0 and marks<=100:  # Data Validation
            self.__marks = marks
            print("Marks Added Successfully")
        else:
            print("Invalid Marks")
    # Display marks
    def get_marks(self):
        return f"Your Current Marks:{self.__marks}"

s = Student()
s.set_marks(85)
print(s.get_marks())
````

## Note:

* Data Validaction
* Controlled Access

### Encapsulation using @property Decorator [Built-in]

````python
class Employee:
    def __init__(self):
        self.__salary = 0

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self,amount):
        if amount>0:
            self.__salary = amount
        else:
            print("Invalid Salary")
````

````python
emp = Employee()
emp.salary = 40000  # setter method
print(emp.salary)  # property method
````

# 2. Method Overriding(Most Important)

* Child calss provides it's own implementation of parent method

````python
class Employee:
    def role(self):
        return "I am an Employee"

class Trainer(Employee):
    def role(self):
        return "I am a Trainer"

class TeamLead(Employee):
    def role(self):
        return "I am a Team Leader"
````

````python
e = Employee()
t = Trainer()
t1 = TeamLead()

print(e.role())
print(t.role())
print(t1.role())
````

````python
t = Trainer()
t.role()  # overrind method role on child and it show single role 
````

Notes:

* Same Method role()
* Different outputs
* Based on Object type
* Runtime Time Polymorphism
