## 1. Object-Oriented Programming structure

##### What is OOP?

`OOP` is a programming paradigm(An OOP paradigm is a style of coding that builds software around objects instead of logic) that organized software design around objects, rather than function and logic.

### Why OOP?
* Code Reusebility.
* Modularity: [Modularity in Object-Oriented Programming (OOP) is the practice of breaking a complex software system into smaller, independent, and self-contained units]
* Scalability: [Scalability in OOP refers to a system's ability to seamlessly handle growth—whether it's increased traffic, data volume, or new features—without crumbling]
* Easy Maintaince
* Real-World Modeling

## 2. Class:

A `class` is a blueprint or template used to create objects.


### Syntax:

```python
class Student:
    pass
```


### Keypoints

* Define using class keyword
* Can contains variable, methods and attribute.
* Name Convention: PascalCase


````python
class Student: # class is keyword and Student is the name class.
    def display(self):  # Display is method and self is keyword 
        print("Welcome to Python OOP") # this is statemnet 
````

````python
avnish = Student()  # avnish is object and Student() is class name
avnish.display()  # calling mehtod using object avnish
````

### 4. `self` Keyword

* `self` refers to the current object of the class.

### Why Required?
* To access instance variable and Methods
* Differntiates instance vairables from local variables.

### Example:

```python
class Demo:
    def show(self):
        print("Self represents current object")
```

````python
class Demo:
    def show(self):
        print("Self represents current object")
````

````python
obj = Demo()
obj.show()
````

### 5. Constructor(`__init__`)

<!-- int a; # variable declaration
a = 32 # variable initialization -->

A `construictor initializes` object data automatically when the object is created.

### syntax

```python
def __init__(self):
```

##### Example

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

s = Student("Alice",22)
```

### Keypoints:

* Automatically invoked
* Used to initialize instance variables.

### Example:

````python
class Student:
    collegeName = "IIT Roorkee" # class varible
    def __init__(self,name,age):
        self.name = name
        self.age = age
````

````python
s1 = Student("Avnish",99)
````

### 7. Types of Methods

1. Instance Mehtods
   * Uses self keyword
   * Accessess instance variable
3. Class Methods
   * Uses [In-Built Decorator: @classmethod]
   * Uses cls
   * Example
```python
class Student:
    collegeName = "IIT Roorkee" # class varible

    @classmethod
    def classMethod(cls):
        print(f"{cls.collegeName}")
```
5. Static Method
   * Uses [In-Built Decorator: @staticmethod]
   * No self or cls
```python
class Student:
    
    @staticmethod
    def info():
        print("Information about staticmethod")
```
