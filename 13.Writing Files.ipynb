## 13. Writing Files in Python

Q. What is file Writing?

`Writing files` means storing data permanently in a file from a Python program.

📌 Used for:
* Logs
* Reports
* Exporting result
* Backup files

### File Writing Modes

|Mode|Meaning|
|----|----|
|`w`|Write(Created new/Overwrites)|
|`a`|append(add data at end)|
|`w+`|Write+Read|
|`a+`|Append and Read|
|`x`|Create if already exists it show error|

----
### Step-by-step process
1. Open the file
2. Write data in the file
3. Close the file

```python
fileName = "C:\\Users\\User\\Desktop\\amough.txt"

# step1
file = open(fileName,'w')
# step - 2
file.write("Myself Amogh\n")
file.write("Samosa is my Favorite Dessert.")

# step-3
file.close()
```

```python
fileName = "C:\\Users\\User\\Desktop\\amough.txt"

# step1
file = open(fileName,'a')
# step - 2
file.write("\nMyself Avnish\n")
file.write("Samosa is my Favorite Dessert.")

# step-3
file.close()
```

```python
fileName = "C:\\Users\\User\\Desktop\\avnish.txt"

# step1
file = open(fileName,'x')
# step - 2
file.write("\nMyself Avnish\n")
file.write("Samosa is my Favorite Dessert.")

# step-3
file.close()
```

## 2. Writing using writeslines()
##### Multiple Lines 

```python
lines = ["Python\n","Java\n","C++\n","Kotlin\n"]

file = open("language.txt",'w')
file.writelines(lines)
file.close()
```

## 3. Best Practice: Writing using `with`



```python
with open("amogh.txt",'w') as file:
    file.write("Using with Statement\n")
    file.write("File auto closed.")
```

## 4. Writing User Input to File

```python
name = input("Enter your name: ")
age = input("Enter your age: ")


with open("amogh.txt",'w') as file:
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
```

```python
with open("amogh.txt",'w') as file:
    name = input("Enter your name: ")
    age = input("Enter your age: ")
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
```

```python
numbers = [10,20,30,40,50]

with open("numbers.txt","w") as file:
    for num in numbers:
        file.write(str(num)+"\n")
```

### 5. Write + Read(W+ Mode)

```python
with open("log.txt",'w+') as file:
    file.write("Hello world")
    file.seek(0)  # set pointer 0
    print(file.read())
```

### 6. Writing using Loop(Log File)


```python
with open("log.txt",'w') as file:
    for i in range(3):
        file.write(f"Log entry:{i}\n")
```

### 7. Exception handling While Writing



```python
try:
    with open('data.txt','x') as file:
        file.write("Hello world")
except Exception as e:
    print("Error",e)
```

### PART-2: CSV File writing in Python

##### What is CSV?
* Comma Seperated Value File
* Used in Excel, databse, reports

##### Required Library
`import csv`

```python
import csv
```

```python
with open("student.csv",'w',newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["ID","Name","Marks"])
    writer.writerow([1,"Ram",99])
    writer.writerow([2,"Ramesh",85])
    writer.writerow([3,"Geeta",78])
```

Notes:

* ✔newline = "" avoids blank line
* ✔csv used in data Science and bacnkend
* ✔Very important for real projects.

```python
wb = Workbook()

sheet = wb.active

#### create column in excel file
sheet["A1"] = "ID"
sheet["B1"] = "Name"
sheet["C1"] = "Salary"

### Add Data in column
sheet.append([1,"Ram",90000])
sheet.append([2,"Ramesh",45000])
sheet.append([3,"Shyam",38000])

wb.save("employee.xlsx")
```
