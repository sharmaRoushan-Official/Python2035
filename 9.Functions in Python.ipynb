### 1. Positional Arguments



```python
def student(name,age):
    print(name, age)

student("John",32)  # correct position
```

```python
def student(name,age):
    print(name, age)

student(32,"John")  # Positional Arguemnt but Wrong position
```

## 2. Keyword Arguments



```python
def student(name,age):
    print(name,age)
student(name="John",age=32)  # keyword arguments
```

```python
def student(name,age):
    print(name,age)
student(age=32, name="John") # keyword arguemnts
```

### 3. Default Arguments


```python
def greet(name="Guest"): # name is default parameter and Default value is Guest
    print("Hello ",name)
greet()  # no arugment it will return default value
```

```python
greet("Amogh")  # if you pass arguemnt then arguemnted value will be returned.
```

### 4. Variable length Arguments

`* args`

`** kwargs`

```python
def total(args):  # args means arguments
    print(sum(args)) # value sum of value
total(10) # sum takes collection data type: list, tuple, set,
```

```python
def total(args):  # args means arguments
    print(sum(args)) # value sum of value
total([10]) # we cann't pass list directly as parameter
```

```python
def total(*args):  # args means arguments
    print(sum(args)) # value sum of value
total(10) # one argument
```

```python
total(10,20) # two arugments
```

```python
total(10,20,30)  # three arguemnts
```

```python
total(10,20,30,40) # four arguments
```

```python
def total(*num):  # args means arguments
    print(sum(num)) # value sum of value
total(10) # one argument
```

```python
total(10,20,50,60,70,80,90,40,1,22,522,522,52) # variable lenght arguments
```

### 7. Scope of Variable 

1. Local Scope Variable
2. Global Scope Variable

### 1.Local Scope Variable

```python
def test():
    x1 = 10  # x1 is a local variable : beacuse x1 is in the block of function 
    print(x1)
test()  # directly value calling
```

```python
x = test()  # x is object through value calling
x
```

```python
print(x1)  # Here, x1 is calling outside the function block then it throw error
```

```python
num = 10 # Global variables [inside + outside]

def callingFunctionValue():
    name = "Harry" # name is local variable it cannot accesssible outiside the function
    print("Calling Outside Variable:",num) # num variable calling inside function then it display value because it is global

callingFunctionValue()
```

```python
print(num)  # calling outside the function it displayed the value
```

### global keyword 

```python
def callingFunctionValue():
    global name # name is local then when you put global keyword before variable it created local to global scope
    name = "Harry" # name is local variable it cannot accesssible outiside the function
    print(name)

callingFunctionValue()
```

### 8. Recursion:
###### One variable calling itself again and again

### Function calling iteself

```python
def factorial(n):  # 3
    if n == 1:
        return 1
    return n*factorial(n-1) # 3 * 2 = 6 ,  3*

factorial(5)
```

3! = 1,2,3 = 6
5! = 1,2,3,4,5=120

### 9. Nested Functions
Function inside another function

```python
def outer():
    def inner():
        print("Inside inner function")
    inner()
outer()
```

```python
def outer():
    def inner():
        print("Inside inner function")
    inner()
    return "Hello world"
outer()
```

### 11. DocString

```python
def add(a,b):
    """
    This function returns sum of two numbers:
    """
    return a+b

print(add.__doc__)
add(10,20)
```

### 13. Annotations(Type Hint)

```python
def add(a,b):
    return a+b

add(10,20)  # integer value return sum integer
```

```python
add('Hello'," World")  # string return string
```

```python
def add(a:int, b:int) -> int:
    return a+b
```

```python
add(10,20)
```

```python
add('Hello'," World")  # string return string
```
