1.Accepts a string froms the user and store each charater in a list.

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, ]
common_elements = list(set(list1) & set(list2))
print(common_elements)
```

4.Convert a tuple into a string.

```python
my_tuple = ('apple', 'banana', 'cherry')
result = str(my_tuple)
print(result) 
print(type(result))
```

5.Count a number of words in a sentence using string method and list.

```python
sentence = "Count a number of words in a sentence"
word_list = sentence.split()
word_count = len(word_list)
print(f"List of words: {word_list}")
print(f"Word count: {word_count}")
```
