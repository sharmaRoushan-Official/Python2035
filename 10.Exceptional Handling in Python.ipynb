## 10. Exceptinal Hanlding in Python 


### 1. `Exception handling` in Python is used to __handle runtime errors__ so that the program does not crash suddenly and can continue execution smoothly.


### 2. Types of Errors

1. Compile Time Error - Code Conversion --> [Highlevel Language --> Lowlevel Language] ---> User Can Fix
2. Run Time Error - Code Execution --> When your Executed by CPU.



# Types of Errors

| Error Type | Definition |
|:-----------|:-----------|
| **Compile Time Error** | An error detected during code conversion (**High-Level → Low-Level Language**) that prevents the program from compiling, and must be fixed by the user before execution. |
| **Run Time Error** | An error that occurs during code execution (when the CPU actually runs the compiled instructions), even though the code compiled successfully. |

> ### 📝 Quick Note
>
> - **Compile Time Error** → Caught by the **compiler/interpreter translation phase**.
>   - **Examples:** Syntax errors, missing semicolons (`;`), type mismatches.
>
> - **Run Time Error** → Caught only **while the program is running**.
>   - **Examples:** Division by zero, null reference, array index out of bounds.

### 1. What is an Exception?

An `exception` is an unexcepted event/error that occurs during the execution of a program.

Example

````python
a = 10
b = 0

print(a/b)
print("Hello world")
````

Note: Here the progam crashes. This is called an exception.

### 3. Why Exception Handling is Required?

Without exception handling:

 * program crashed
 * user gets error messages
 * system becomes unreliable. [unreliable means you cannot be trusted or depended upon to do what is expected]

With exception handling:

 * progam continues
 * we can show user-friendly messages
 * better debugging and loggin.
 * prevents data loss

## 7. Multiple Except Blocks

You can handle multiple errors

Example

````python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    print(x/y)

except ZeroDivisionError:
    print("Cannot divide by zero")
except ValueError:
    print("Invalid Input! Please enter a valid number.")
````

## 8. Single Except Handling Multiple Exceptions

Syntax:

except(Exception1,Exception2)

````python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    print(x/y)

except (ZeroDivisionError,ValueError):
    print("Cannot divide by zero OR Invalid Input")  # this statement runs on both errors
````

## 9. Using Exception Object(as e)  as-alias

This is used to print actual error message.

````python
# Recommended and Best of all errors
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    print(x/y)

except Exception as e:
    print("Error Occurred:",e)
````

### 10. try-except-else Block

else block runs only if no exception occurs.

Example

````python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    result = x / y

except Exception as e:  # e is the object of error
    print("Error Occurred:",e)

else:
    print(f"The Division of two numbers {x} and {y} is: {result}")
````

Note:

Flow:

* if exception ---> except runs
* if no exception --> else runs

### 11. try-except-else-finally

Finally block always runs(whether exception happens or not.)

````python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    result = x / y

except Exception as e:  # e is the object of error
    print("Error Occurred:",e)

else:
    print(f"The Division of two numbers {x} and {y} is: {result}")

finally:
    print("Program Excution finished.")
# print("Hello world")
````

### 13. Nested try-except

When exception handling is needed inside antoher tyr.

````python
try:
    a = int(input("Enter A: "))
    b = int(input("Enter B: "))
    try:
        print(a/b)
    except Exception as e:
        print("Inner Block Error:",e)
except Exception as e:
    print("Outer Block Error:",e)
    
````

## 17. Custom Exceptions (User-Defined Exceptions)

You can create your own exception class



````python
class InvalidAgeError(Exception):
    pass
````

````python
age = int(input("Enter age: "))

if age < 18:
    raise InvalidAgeError("Age must be 18 or above")  # Uisng User defined Error

else:
    print("You Are Elegible")
````
