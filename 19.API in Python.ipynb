### 3. HTTP Methods

This most important methods

|Method|Purpose|
|----|----|
|`GET`| Retrieve data|
|`POST`| Create/send data|
|`PUT`| Update Complete data|
|`PATCH`| Update partial data|
|`DELETE`| Delete data|

Example

```
GET  ➛ Get Student
POST ➛ Create Student
PUT ➛ Update Student
PATCH ➛ Update some fields
DELETE ➛ Delete Student

### 5. Calling API using Python

The most commonly taught library is `requests`

Install 

IN CMD:

`pip install requests`



````python
import requests

url = "https://jsonplaceholder.typicode.com/users"
print(url)

response = requests.get(url)

print(response.status_code)
print(response.text)
````

### 6. Convert API Response to Python

Instead of .text, use .json()

````python
import requests

url = "https://jsonplaceholder.typicode.com/users"
print(url)

response = requests.get(url)

print(response.status_code)
# print(response.text)
data = response.json()
print(data)
````

Notes:

Not data is a Python list/dictionary structure.

## 7. Access API Data



````python
import requests

url = "https://jsonplaceholder.typicode.com/users"
print(url)

response = requests.get(url)

users = response.json()  # dictionary

for user in users:
    print("ID:",user["id"],"\nName:",user["name"],"\nUser Name:",user["username"])
    print("\n")
````
