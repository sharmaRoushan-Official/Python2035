### Woriking with JSON in Python

Python provides built-in module:-

`import json`

```python
import json
```

1. Convert Python  --> JSON

```python
data = {
    "name":"Avanish",
    'age':65,
    "course":"Python Programming"
}
print(data,type(data))
```

Python Dcitonary to JSON String

`dumps:Python dictionary-->JSON`

```python
json_data = json.dumps(data)
print(json_data)  # string
```

### 2. convert JSON --> Python Object

`json.loads()`

```python
json_data1 = '{"name": "Avanish", "age": 65, "course": "Python Programming"}'
print(json_data1,type(json_data1))

python_dict = json.loads(json_data1)
print(python_dict,type(python_dict))
```

### 3. Working with JSON Files

`dump() --- Python dictionary to JSON File`

```python
data = {
    'name':"Amogh",
    "age":45
}
print(data,type(data))
```

```python
with open("data.json",'w') as file:
    json.dump(data,file,indent=4)
```

### 4. Reading JSON form File

`load() ---> JSON File to Python dictionary`

```python
with open("data.json","r") as file:
    data = json.load(file)
    print(data)
```

### dump() vs dumps() and load() vs loads()


|Function|Use|
|---|---|
|dump()|write to file|
|dumps()|convert to string|
|load()|Read from file|
|loads()|Convert string to dictionary|
