# 16. Decorator in Python

Q. What is Decorator?

A `decorator` is a function that modifies another function's behavior without changing its code.

📌Define
> A decorator adds extra feature to a function.

##### Why Decorators are Needed
❌ Without Decorators

if you want to:
* Log function call
* Check user authentication
* Measure execution time.

✅️ With Decorator
* Write the logic once
* Reuse it many times.
* Cleaner and maintaintable code.


### Real-Life Analogy

Gift = Original Gift

Wrapper =  Decorator

Gift remains same, Wrapper adds value

Function are First-Class and object



### Applying a Decorator Manual way

```python
def decorator(func):
    def wrapper():
        print("Before Function")
        func() # say_hello()
        print("After Function")
    
    return wrapper


def say_hello():
    print("Hello world")


say_hello = decorator(say_hello)

say_hello()
```

## using @ Syntax [Recommended]



```python
@decorator
def say_hello():
    print("Hello world")
say_hello()
```

### Decorator with *args and **kwargs

```python
def decorator(func):
    def wrapper(name):
        print("Congratulations!!")
        func(name)
    return wrapper
```

```python
def greet(name):
    print("Hello",name)
greet("Amough")
```

```python
# manual calling
val = decorator(greet)
val("Amogh")
```

```python
# @ syntax
@decorator
def greet(name):
    print("Hello",name)
greet("Amough")
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function Started")
        result = func(*args,**kwargs)
        print("Function Ended")
        return result
    return wrapper
```

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        print("Function Started")
        start = time.time() # start timing
        result = func(*args,**kwargs)
        end = time.time() # end time

        exeTime = end-start
        print("Execution Time:",exeTime)
        print("Function Ended")
        return result
    return wrapper
```

```python
@timer
def demo():
    for i in range(100000000):
        pass
demo()
```

### Multiple Decorators



```python
def decor1(func):
    def wrapper():
        print("Decorator 1")
        func()
    return wrapper


def decor2(func):
    def wrapper():
        print("Decorator 2")
        func()
    return wrapper
```

```python
@timer
@decor2
@decor1
def show():
    return "Hello world"
show()
```
